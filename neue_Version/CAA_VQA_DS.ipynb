# Create VQA dataset from CLEVR

## code

```python
import json
import random

def prepare_counting_extraction_dataset(
    clevr_questions_path: str, 
    output_jsonl_path: str, 
    num_pairs: int = 1000,
    train = False
):
    """
    Parses the CLEVR questions JSON and generates a contrastive dataset 
    for extracting a '+1 Overcount' steering vector.
    """
    print(f"Loading CLEVR questions from {clevr_questions_path}...")
    with open(clevr_questions_path, 'r') as f:
        clevr_data = json.load(f)
    
    questions = clevr_data['questions']    
    valid_numbers = {str(i) for i in range(10)}
    contrastive_pairs = []
    
    print("Filtering and generating contrastive pairs...")
    for q in questions:
        true_answer = str(q['answer'])
        if true_answer in valid_numbers:
            contrastive_pairs.append({
                "image_path": q['image_filename'],
                "question": q['question'],
                "true_answer": true_answer,
                "target_answer": str(int(true_answer) + 1) if train else true_answer
            })

    random.shuffle(contrastive_pairs)
    if len(contrastive_pairs) < num_pairs:
        print(f"Warning: Only found {len(contrastive_pairs)} valid counting questions.")
    else:
        contrastive_pairs = contrastive_pairs[:num_pairs]

    print(f"Writing {len(contrastive_pairs)} pairs to {output_jsonl_path}...")
    with open(output_jsonl_path, 'w') as f:
        for pair in contrastive_pairs:
            f.write(json.dumps(pair) + '/n')
            
    print("Dataset preparation complete!")
```

## train split

```python
CLEVR_TRAIN_JSON = "/kaggle/input/datasets/timoboz/clevr-dataset/CLEVR_v1.0/questions/CLEVR_train_questions.json"
OUTPUT_FILE = "clevr_caa_extract_train10.jsonl"

prepare_counting_extraction_dataset(
    clevr_questions_path=CLEVR_TRAIN_JSON,
    output_jsonl_path=OUTPUT_FILE,
    num_pairs=10
)
```
